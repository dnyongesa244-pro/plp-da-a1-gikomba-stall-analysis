```python
#These are the input data recieved from wanjiku sales
daily_sales = [2300,1950, 2800, 2100, 3600, 5200, 4100, 2500, 2200, 3100, 2400, 3900, 5600, 4700]
days = ["Mon", "Tue", "Wed","Thu","Fir","Sat","Sun"] * 2
#qty = kg in stock, reorder = reorder level(kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":{"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":{"qty": 9,"reorder": 15, "cost": 30, "price": 50},
    "Potatoes":{"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":{"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}
```

```python
#Task 1
total = sum(daily_sales)  #This adds all the daily sales from Wanjikus record data
avarage = total/len(daily_sales)  #The avarage is dound by taking the total_sales/sales_count_days wich is the lenth of the sales_list
highest_sales = max(daily_sales) #We use max(list) to find the highest sales out of the 14 days
lowes_sales = min(daily_sales) #we use min(list) to find the lowes sales out of the 14 days

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowes_sales)

def week_no(day):
    if day<=6 :
        return 1
    else:return 2    
best_week = week_no(best_index)
worst_week = week_no(worst_index)


print(f"Total sales : {total}")
print(f"Avarage sale : {avarage}")
print(f"Higest sale : {highest_sales}")
print(f"Lowest sale : {lowes_sales}")
print(f"Best day : {days[best_index]} of week {best_week} ({highest_sales} KES)")
print(f"Worst day : {days[worst_index]} of week {worst_index} ({lowes_sales} KES)")
```

```python
#Task 2
def week_totals(sales):
    week_one_total = sum(sales[0:7]) #find the week one total stating from 0 to 6th index
    week_two_total = sum(sales[7:])  #find the week two total stating from 7th to last index
    return week_one_total, week_two_total  #returns the week one and week 2 total

week1, week2 = week_totals(daily_sales)
print(f"Week 1 total : {week1:,}  KES")
print(f"Week 2 total : {week2:,}  KES")
percentage_growth = (24400-22050)/22050*100
print(f"percentage growth : {percentage_growth:,.2f}")
```

```python
#Task 3
def clasify_day(amount, avarage):
    percent = (amount-avarage)/avarage*100
    if percent > 20:
        return "Strong"
    elif percent <-20:
        return "Weak"
    else:
        return "Normal"

counts = {"Strong": 0, "Normal" : 0, "Weak" : 0}

for amount in daily_sales:
    label = clasify_day(amount, avarage)
    counts[label]+=1


for label, count in counts.items():
    print(f"{label} : {count} days")               
```

```python
week_one = daily_sales[0:7]
week_two = daily_sales[7:]

weekday_sales = week_one[0:5]+week_two[0:5]
weekend_sales = week_one[5:]+week_two[5:]
weekday_avg = sum(weekday_sales)/len(weekday_sales)
weekend_avg = sum(weekend_sales)/len(weekend_sales)


difference_pct = (weekend_avg-weekday_avg)/weekday_avg*100

print(f"Weekday avarage : {weekday_avg}")
print(f"weekend avarage : {weekend_avg}")
print(f"Weekend is {difference_pct:,.2f} higher than weekday")
```

```python
def items_to_reorder(stock):
    low_stock = []
    for name, info in stock.items():
        if info["qty"] <= info["reorder"]:
            low_stock.append(name)
    return low_stock

def gross_margins(stock):
    margins = {}
    for name, info in stock.items():
        margins[name] = (info["price"] - info["cost"]) / info["price"] * 100
    return margins

print("Reorder:", items_to_reorder(stock))

for name, margin in gross_margins(stock).items():
    print(f"{name}: {margin:.1f}%")
```
