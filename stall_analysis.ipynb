# Gikomba Stall Analysis

An analysis of Wanjiku's 14-day sales and current stock to understand performance, classify sales days, and make restocking recommendations.

```python
#These are the input data recieved from wanjiku sales
daily_sales = [2300,1950, 2800, 2100, 3600, 5200, 4100, 2500, 2200, 3100, 2400, 3900, 5600, 4700]
days = ["Mon", "Tue", "Wed","Thu","Fir","Sat","Sun"] * 2
#qty = kg in stock, reorder = reorder level(kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":{"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":{"qty": 9,"reorder": 15, "cost": 30, "price": 50},
    "Potatoes":{"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":{"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}
```

## Task 1: Summarise sales

I calculate total, average, highest, and lowest sales, then use the day index to identify when the best and worst sales occurred.

```python
#Task 1
total = sum(daily_sales)  #This adds all the daily sales from Wanjikus record data
avarage = total/len(daily_sales)  #The avarage is dound by taking the total_sales/sales_count_days wich is the lenth of the sales_list
highest_sales = max(daily_sales) #We use max(list) to find the highest sales out of the 14 days
lowes_sales = min(daily_sales) #we use min(list) to find the lowes sales out of the 14 days

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowes_sales)

def week_no(day):
    if day<=6 :
        return 1
    else:return 2    
best_week = week_no(best_index)
worst_week = week_no(worst_index)


print(f"Total sales : {total}")
print(f"Avarage sale : {avarage}")
print(f"Higest sale : {highest_sales}")
print(f"Lowest sale : {lowes_sales}")
print(f"Best day : {days[best_index]} of week {best_week} ({highest_sales} KES)")
print(f"Worst day : {days[worst_index]} of week {worst_index} ({lowes_sales} KES)")
```

## Task 2: Compare weekly sales

I define a function to total each seven-day week and calculate the percentage change from week 1 to week 2.

```python
#Task 2
def week_totals(sales):
    """Return the sales totals for the first and second seven-day weeks."""
    week_one_total = sum(sales[0:7]) #find the week one total stating from 0 to 6th index
    week_two_total = sum(sales[7:])  #find the week two total stating from 7th to last index
    return week_one_total, week_two_total  #returns the week one and week 2 total

week1, week2 = week_totals(daily_sales)
print(f"Week 1 total : {week1:,}  KES")
print(f"Week 2 total : {week2:,}  KES")
percentage_growth = (24400-22050)/22050*100
print(f"percentage growth : {percentage_growth:,.2f}")
```

## Task 3: Classify sales days

I compare each day's sales with the overall average, label it Strong, Normal, or Weak, and count the days in each category.

```python
#Task 3
def classify_day(amount, average):
    """Classify sales as Strong, Normal, or Weak relative to the average."""
    percent = (amount-average)/average*100
    if percent > 20:
        return "Strong"
    elif percent <-20:
        return "Weak"
    else:
        return "Normal"

counts = {"Strong": 0, "Normal" : 0, "Weak" : 0}

for amount in daily_sales:
    label = classify_day(amount, avarage)
    counts[label]+=1


for label, count in counts.items():
    print(f"{label} : {count} days")               
```

## Task 4: Compare weekdays and weekends

I separate the two weeks into weekday and weekend sales, calculate each average, and measure how much higher weekend sales are.

```python
#Task 4
week_one = daily_sales[0:7]
week_two = daily_sales[7:]

weekday_sales = week_one[0:5]+week_two[0:5]
weekend_sales = week_one[5:]+week_two[5:]
weekday_avg = sum(weekday_sales)/len(weekday_sales)
weekend_avg = sum(weekend_sales)/len(weekend_sales)


difference_pct = (weekend_avg-weekday_avg)/weekday_avg*100

print(f"Weekday avarage : {weekday_avg}")
print(f"weekend avarage : {weekend_avg}")
print(f"Weekend is {difference_pct:,.2f} higher than weekday")
```

## Task 5: Review stock and margins

I identify products at or below their reorder levels and calculate each product's gross margin from its cost and selling price.

```python
#Task 5
def items_to_reorder(stock):
    """Return product names whose quantities are at or below reorder levels."""
    low_stock = []
    for name, info in stock.items():
        if info["qty"] <= info["reorder"]:
            low_stock.append(name)
    return low_stock

def gross_margins(stock):
    margins = {}
    for name, info in stock.items():
        margins[name] = (info["price"] - info["cost"]) / info["price"] * 100
    return margins

print("Reorder:", items_to_reorder(stock))

for name, margin in gross_margins(stock).items():
    print(f"{name}: {margin:.1f}%")
```

## Task 6: Print the performance report

I bring together the sales, growth, day-type, and stock results in a formatted report, using KES amounts with thousands separators.

```python
#Task 6
def kes(amount):
    return f"KES {amount:,.2f}"

best_index = daily_sales.index(max(daily_sales))
worst_index = daily_sales.index(min(daily_sales))
week1, week2 = week_totals(daily_sales)
growth = (week2 - week1) / week1 * 100
margins = gross_margins(stock)

line = "=" * 44
print(line)
print("STALL PERFORMANCE REPORT".center(44))
print("Wanjiku, Gikomba Market (14 days)".center(44))
print(line)

print("SALES")
print(f"  {'Total sales':<22}{kes(total):>20}")
print(f"  {'Average per day':<22}{kes(avarage):>20}")
print(f"  {'Best day':<22}{days[best_index] + ' wk ' + str(week_no(best_index)):>12} {kes(max(daily_sales)):>12}")
print(f"  {'Worst day':<22}{days[worst_index] + ' wk ' + str(week_no(worst_index)):>12} {kes(min(daily_sales)):>12}")

print("\nGROWTH")
print(f"  {'Week 1':<22}{kes(week1):>20}")
print(f"  {'Week 2':<22}{kes(week2):>20}")
print(f"  {'Week-on-week':<22}{growth:>19.2f}%")

print("\nWEEKDAY vs WEEKEND")
print(f"  {'Weekday average':<22}{kes(weekday_avg):>20}")
print(f"  {'Weekend average':<22}{kes(weekend_avg):>20}")
print(f"  {'Weekend is higher by':<22}{difference_pct:>19.1f}%")

print("\nDAY TYPES")
for label, count in counts.items():
    print(f"  {label:<22}{count:>14} days")

print("\nSTOCK")
print(f"  {'Product':<12}{'Qty':>5}{'Reorder':>9}{'Margin':>9}  Status")
for name, info in stock.items():
    status = "RESTOCK" if info["qty"] <= info["reorder"] else "OK"
    print(f"  {name:<12}{info['qty']:>5}{info['reorder']:>9}{margins[name]:>8.1f}%  {status}")
print(line)
```

## Task 7: Recommend actions

I use the sales trends, weekday/weekend comparison, and inventory margins to recommend practical actions for Wanjiku.

## Recommendations for Wanjiku

3. **Your business is growing, but keep tracking**: sales rose from 22,050 KES in week 1 to 24,400 KES in week 2 (+10.66%), though two weeks is too short to call it a lasting trend.
4. **Focus on Avocado and Sukuma for profit**: they have the best margins (42.9% and 40.0%), so each kilo sold earns more than Onions (30.0%), and Avocado (12 kg vs reorder level 10) is close to needing a top-up.
5. **Look for ways to lift weekdays**: 6 of the 14 days were Weak, and the weekday average of 2,685 KES is well below the overall average of 3,317.86 KES.
